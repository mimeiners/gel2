# Einstieg in LTspice

Sie benötigen eine Bibliothek aus dem Python-Paket-Archiv, um mit LTSpice interagieren zu können. Entnehmen Sie die
notwendigen Installationsschritte bitte der entsprechenden Webseite [PyLTSpice](https://github.com/nunobrum/PyLTSpice).

## Laden der notwendigen Bibliotheken

````python
from PyLTSpice import RawRead
import matplotlib.pyplot as plt
````

## Einlesen der Rohdaten

````python
ltr = RawRead('../files/spice/bsp433_reisch.raw')
# print(ltr.get_trace_names())
# print(ltr.get_raw_property())
````

## Zuweisen der Daten zu Variablen

````python
t = ltr.get_trace('time')
V1 = ltr.get_trace('V(1)')
V2 = ltr.get_trace('V(2)')
````

## Plotten der Spannungsverläufe

````python
#| label: fig-rc-tran
#| fig-cap: "Transienten am RC-Glied"
#| code-fold: false
#| enabled: true

plt.plot(t, V1, label="$v_1(t)$")
plt.plot(t, V2, label="$v_2(t)$")
plt.legend()
plt.grid()
plt.xlabel('Zeit t in s')
plt.ylabel('Spannung V in V')
plt.show()
````
